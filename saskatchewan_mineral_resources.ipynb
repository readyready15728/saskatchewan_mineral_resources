```python
import folium
import geopandas as gpd
from folium.plugins import BeautifyIcon
```

To begin with, the dataset came from: https://geohub.saskatchewan.ca/datasets/saskatchewan::mineral-deposits-index/about. After loading the dataset, I examined the CRS to make sure it's WGS 84, which is required by Folium. Being consciously aware of the CRS and its strengths and weaknesses is always important for GIS tasks!

```python
gdf = gpd.read_file('Mineral_Deposits_Index.geojson')
gdf.crs
```

Let's take a quick look at the data:

```python
gdf.head()
```

Most of the columns are unnecessary for what I set out to do. Here's a look at a much more minimal dataset:

```python
columns_to_drop = [
    'OBJECTID',
    'SMDI',
    'HISTORICNAMES',
    'ASSOCIATEDCOMMODITIES',
    'GROUPING',
    'DISCOVERYTYPE',
    'STATUS',
    'PRODUCTION',
    'RESERVESRESOURCES',
    'UTM13E',
    'UTM13N',
    'SYMBOLOGY_STATUS',
    'SYMBOLOGY_GROUPING',
    'GLOBALID'
]

gdf = gdf.drop(columns_to_drop, axis=1)
gdf.head()
```

Let's rename all columns except `geometry` for easier handling:

```python
renamed_columns = {
    'NAME': 'name',
    'PRIMARYCOMMODITIES': 'primary_commodities',
    'WEBLINK': 'weblink'
}

gdf = gdf.rename(columns=renamed_columns)
gdf.head()
```

The following code looks at each entry under `primary_commodities`, splits and strips any comma-separated values, and enters them into a dictionary where the keys will be each commodity as used in the following code and the value is as it will appear in a more human-readable format in the map legend, as it was originally:

```python
def convert_label_to_key(label):
    label = label.lower()
    label = label.replace('(', '')
    label = label.replace(')', '')
    label = label.replace(' ', '_')

    return label

commodities_with_labels = {}

for row in gdf['primary_commodities']:
    commodities = [commodity.strip() for commodity in row.split(',')]

    for commodity in commodities:
        commodities_with_labels[convert_label_to_key(commodity)] = commodity
```

Here are the results:

```python
for k, v in sorted(commodities_with_labels.items(), key=lambda pair: pair[0]):
    print(f'{k}: {v}')
```

Here the `GeoDataFrame` will be made much wider by creating a column for each commodity with `True` if the commodity was present at the site corresponding to the row in question and `False` otherwise. (I was most easily able to inspect the widened results by saving the `GeoDataFrame` to a temporary `.csv` file and looking at it in Gnumeric.) The result is definitely not "tidy", but untidy data might be easier to work with in Python than in R with Tidyverse packages:

```python
for key in sorted(commodities_with_labels):
    gdf[key] = gdf['primary_commodities'].apply(lambda row: commodities_with_labels[key] in row)
```

`primary_commodities` is no longer necessary and I prefer `geometry` to be moved to the end of the columns (though GeoPandas does not require it to be there):

```python
gdf = gdf.drop('primary_commodities', axis=1)
gdf = gdf.assign(geometry=gdf.pop('geometry'))
gdf.head()
```

At last the map will be ready, including hyperlinks that lead to the appropriate page for each location on the Saskatchewan government's website. I was able to get the "radio button" functionality (i.e. only one mineral at a time) to work properly with the help of the following Stack Overflow thread:

https://stackoverflow.com/questions/61345801/featuregroup-layer-control-in-folium-only-one-active-layer/63189269#63189269

```python
m = folium.Map(location=[gdf.geometry.y.mean(), gdf.geometry.x.mean()], zoom_start=5, tiles=None)
base_map = folium.FeatureGroup(name='Basemap', overlay=True, control=False)
folium.TileLayer(tiles='OpenStreetMap').add_to(base_map)
base_map.add_to(m)

feature_groups = {}

for key in sorted(commodities_with_labels):
    feature_groups[key] = folium.FeatureGroup(commodities_with_labels[key], show=False, overlay=False).add_to(m)

for idx, row in gdf.iterrows():
    for key in commodities_with_labels:
        if row[key]:
            folium.Marker(
                location=[row.geometry.y, row.geometry.x],
                tooltip=row['name'],
                popup='<a href="%s">%s</a>' % ((row['weblink'],) * 2),
                icon=BeautifyIcon(background_color=custom_colors.get(key, '#928E85')) # Defaults to "stone gray"
            ).add_to(feature_groups[key])

folium.LayerControl().add_to(m)

m.save('index.html')

m
```
